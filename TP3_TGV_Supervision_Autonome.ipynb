## 🌍 Contexte et introduction  

Avec l’évolution des technologies ferroviaires, les trains modernes ne sont plus de simples moyens de transport mécaniques : 
ce sont de véritables **systèmes cyber-physiques complexes**, dotés de capteurs avancés, de calculateurs embarqués et de réseaux de communication à haut débit.  

La SNCF et ses partenaires industriels (Alstom, Siemens, Hitachi…) travaillent actuellement sur le projet **Train Autonome**.  
Un TGV autonome doit être capable de :  

- **Percevoir son environnement** (obstacles, signaux, conditions extérieures) grâce à des capteurs variés :  
  **LIDAR, caméras, radars, GPS/IMU, capteurs thermiques**, mais aussi des capteurs classiques (tachymètre, balises ETCS, sondes de pression et température).  

- **Traiter l’information en temps réel** grâce à un calculateur central (**TCMS – Train Control and Management System**) et à des modules d’**IA embarquée** pour la fusion de données.  

- **Communiquer** avec l’ensemble des sous-systèmes du train via des réseaux internes (**MVB, Ethernet train**) et avec le sol via des réseaux externes (**GSM-R aujourd’hui, FRMCS/5G demain**).  

- **Restituer l’information** au conducteur (s’il existe encore un rôle de supervision humaine), au système de supervision embarqué, et au centre de contrôle à distance via des **plateformes cloud ferroviaires**.  

---

Autrement dit, le TGV autonome fonctionne comme un **organisme vivant** :  
- ses **capteurs** jouent le rôle des organes sensoriels (vue, ouïe, toucher),  
- ses **réseaux** sont le système nerveux,  
- son **calculateur central (TCMS + IA)** agit comme le cerveau,  
- et son **IHM + cloud** jouent le rôle du langage et de la communication externe.

## 📘 Partie 2 – Capteurs modernes du TGV autonome  

### 1. LIDAR (Light Detection and Ranging)  
- **Principe** : émission d’un faisceau laser → mesure du temps de retour → reconstruction 3D de l’environnement.  
- **Rôle** : détection d’obstacles, aide à la navigation.  

### 2. Caméras embarquées  
- **Principe** : vision artificielle, analyse d’images par IA.  
- **Rôle** : lecture signaux, détection obstacles, surveillance.  

### 3. GPS + Inertial Measurement Unit (IMU)  
- **Principe** : GPS (position absolue) + IMU (gyroscope + accéléromètre).  
- **Rôle** : localisation continue et précise, complément ETCS/KVB.  

### 4. Capteurs environnementaux avancés  
- **Radar Doppler** → vitesse instantanée.  
- **Capteurs thermiques/IR** → détection chaleur, sécurité.  
- **Capteurs acoustiques** → analyse bruit (roulements, freins).

## 📘 Partie 3 – Réseaux ferroviaires  


### 🔹 Réseaux internes (dans le train)
- **MVB (Multifunction Vehicle Bus)**  
  - Bus série normalisé (IEC 61375).  
  - Fiable, robuste, faible débit (~1,5 Mbit/s).  
  - Utilisé pour les systèmes critiques : freins, portes, traction.  

- **Ethernet Train (ETB / ETBN)**  
  - Adaptation de l’Ethernet classique.  
  - Haut débit (100 Mbit/s – 1 Gbit/s).  
  - Supporte vidéosurveillance, Wi-Fi passagers, diagnostics.  

### 🔹 Réseaux sol-train (communication avec l’infrastructure)
- **GSM-R (Global System for Mobile – Railway)**  
  - Basé sur la 2G (GSM).  
  - Communication conducteur ↔ régulateur.  
  - Débit limité (~9,6 kbit/s).  
  - Fin de vie prévue vers 2030.  

- **FRMCS (Future Railway Mobile Communication System)**  
  - Basé sur la 5G ferroviaire.  
  - Très haut débit, faible latence.  
  - Support IoT ferroviaire et conduite autonome.  
  - Remplacera GSM-R dans les prochaines années.  

### 🔹 Supervision distante (cloud ferroviaire)
- Transmission des données en temps réel au **centre de contrôle**.  
- Utilisation de réseaux IP sécurisés (fibre, 5G, VPN).  
- Sert à la maintenance prédictive et à la surveillance globale.


# Partie 4 : Mini-projet Tkinter 

## 🎯 Objectif
Créer une **maquette logicielle de supervision d’un TGV autonome**, intégrant plusieurs modules :
- Supervision embarquée (vitesse, portes, température, signalisation).
- Supervision des capteurs modernes (LIDAR, GPS, caméra IA).
- Réseau interne et communication sol-train.
- Énergie et confort passagers.
- Sécurité globale avec une petite IA décisionnelle.



## 🔹 Consignes

2. Ajoutez des couleurs, icônes et alertes visuelles/sonores.
3. Développez au moins **2 modules de supervision** complets par groupe.
4. Complétez la logique d’IA pour afficher l’état global du train.

## 🔹 Modules proposés
- Sécurité (portes, signalisation, LIDAR).
- Réseau interne (MVB/Ethernet).
- Communication sol-train (GSM-R/FRMCS).
- Énergie (consommation, récupération).
- Confort passagers (climatisation, Wi-Fi).
- Supervision distante (cloud).

👉 Chaque groupe doit personnaliser son interface et documenter ses choix.


````python
import tkinter as tk
import random

# Initialisation de la fenêtre principale
root = tk.Tk()
root.geometry("800x600")
root.configure(bg="white")
root.title("Supervision TGV autonome")
root.resizable(False, False)


# =========================
# VARIABLES SIMULÉES
# =========================

vitesse = tk.IntVar(value=0)
porte_fermee = tk.BooleanVar(value=True)
temperature = tk.IntVar(value=20)
signalisation = tk.StringVar(value="Voie libre")
lidar_obstacle = tk.BooleanVar(value=False)
gps_position = tk.StringVar(value="48.8566N, 2.3522E")
reseau_interne = tk.BooleanVar(value=True)
reseau_gsmr = tk.BooleanVar(value=True)
reseau_frmcs = tk.BooleanVar(value=True)
energie = tk.IntVar(value=50)
wifi = tk.BooleanVar(value=True)
etat_global = tk.StringVar(value="✅ OK – circulation normale")


# =========================
# ANALYSE IA
# =========================

def analyse_IA():

    # 🔴 DANGER
    if (not porte_fermee.get()
        or signalisation.get() == "Arrêt"
        or lidar_obstacle.get()
        or vitesse.get() > 350
        or temperature.get() > 35
        or energie.get() < 20):

        etat_global.set("🚨 Danger : départ interdit")

    # 🟠 ALERTE
    elif (vitesse.get() > 300
          or temperature.get() > 25
          or signalisation.get() == "Ralentissement"
          or not reseau_gsmr.get()
          or not reseau_frmcs.get()
          or energie.get() < 50):

        etat_global.set("⚠️ Alerte")

    # 🟢 OK
    else:

        etat_global.set("✅ OK – circulation normale")


# =========================
# MISE À JOUR DES DONNÉES
# =========================

def maj_donnees():

    vitesse.set(random.randint(200, 360))

    temperature.set(random.randint(15, 40))

    porte_fermee.set(
        random.choice([True, True, True, False])
    )

    signalisation.set(
        random.choice([
            "Voie libre",
            "Voie libre",
            "Ralentissement",
            "Arrêt"
        ])
    )

    lidar_obstacle.set(
        random.choice([False, False, False, True])
    )

    gps_position.set(
        f"{round(random.uniform(48, 49), 4)}N, "
        f"{round(random.uniform(2, 3), 4)}E"
    )

    reseau_interne.set(
        random.choice([True, False])
    )

    reseau_gsmr.set(
        random.choice([True, True, True, False])
    )

    reseau_frmcs.set(True)

    energie.set(
        random.randint(30, 100)
    )

    wifi.set(
        random.choice([True, False])
    )

    analyse_IA()

    root.after(2000, maj_donnees)


# =========================
# INTERFACE GRAPHIQUE
# =========================

# Grande zone du haut
frame_top = tk.Frame(
    root,
    bg="blue",
    height=100
)

frame_top.pack(fill="x")


# Titre
tk.Label(
    frame_top,
    text="Supervision TGV autonome",
    fg="white",
    bg="blue",
    font=("Arial", 22)
).pack()


# Grande zone centrale
frame_middle = tk.Frame(
    root,
    bg="white"
)

frame_middle.pack(
    fill="both",
    expand=True
)


# =========================
# LES 4 FRAMES
# =========================

# Frame 1 : supervision embarquée
frame_embarque = tk.Frame(
    frame_middle,
    bg="lightgray"
)

frame_embarque.grid(
    row=0,
    column=0,
    padx=10,
    pady=10
)


# Frame 2 : capteurs
frame_capteurs = tk.Frame(
    frame_middle,
    bg="lightgray"
)

frame_capteurs.grid(
    row=0,
    column=1,
    padx=10,
    pady=10
)


# Frame 3 : réseaux
frame_reseaux = tk.Frame(
    frame_middle,
    bg="lightgray"
)

frame_reseaux.grid(
    row=1,
    column=0,
    padx=10,
    pady=10
)


# Frame 4 : énergie et confort
frame_energie = tk.Frame(
    frame_middle,
    bg="lightgray"
)

frame_energie.grid(
    row=1,
    column=1,
    padx=10,
    pady=10
)


# =========================
# FRAME EMBARQUÉE
# =========================

tk.Label(
    frame_embarque,
    text="SUPERVISION EMBARQUÉE"
).grid(
    row=0,
    column=0,
    columnspan=2
)

tk.Label(
    frame_embarque,
    text="Vitesse (km/h)"
).grid(row=1, column=0)

tk.Label(
    frame_embarque,
    textvariable=vitesse
).grid(row=1, column=1)


tk.Label(
    frame_embarque,
    text="Porte fermée"
).grid(row=2, column=0)

tk.Label(
    frame_embarque,
    textvariable=porte_fermee
).grid(row=2, column=1)


tk.Label(
    frame_embarque,
    text="Température (°C)"
).grid(row=3, column=0)

tk.Label(
    frame_embarque,
    textvariable=temperature
).grid(row=3, column=1)


tk.Label(
    frame_embarque,
    text="Signalisation"
).grid(row=4, column=0)

tk.Label(
    frame_embarque,
    textvariable=signalisation
).grid(row=4, column=1)


# =========================
# FRAME CAPTEURS
# =========================

tk.Label(
    frame_capteurs,
    text="CAPTEURS"
).grid(
    row=0,
    column=0,
    columnspan=2
)

tk.Label(
    frame_capteurs,
    text="Obstacle LIDAR"
).grid(row=1, column=0)

tk.Label(
    frame_capteurs,
    textvariable=lidar_obstacle
).grid(row=1, column=1)


tk.Label(
    frame_capteurs,
    text="Position GPS"
).grid(row=2, column=0)

tk.Label(
    frame_capteurs,
    textvariable=gps_position
).grid(row=2, column=1)


# =========================
# FRAME RÉSEAUX
# =========================

tk.Label(
    frame_reseaux,
    text="RÉSEAUX"
).grid(
    row=0,
    column=0,
    columnspan=2
)

tk.Label(
    frame_reseaux,
    text="Réseau interne"
).grid(row=1, column=0)

tk.Label(
    frame_reseaux,
    textvariable=reseau_interne
).grid(row=1, column=1)


tk.Label(
    frame_reseaux,
    text="GSM-R"
).grid(row=2, column=0)

tk.Label(
    frame_reseaux,
    textvariable=reseau_gsmr
).grid(row=2, column=1)


tk.Label(
    frame_reseaux,
    text="FRMCS"
).grid(row=3, column=0)

tk.Label(
    frame_reseaux,
    textvariable=reseau_frmcs
).grid(row=3, column=1)


# =========================
# FRAME ÉNERGIE ET CONFORT
# =========================

tk.Label(
    frame_energie,
    text="ÉNERGIE ET CONFORT"
).grid(
    row=0,
    column=0,
    columnspan=2
)

tk.Label(
    frame_energie,
    text="Énergie (%)"
).grid(row=1, column=0)

tk.Label(
    frame_energie,
    textvariable=energie
).grid(row=1, column=1)


tk.Label(
    frame_energie,
    text="Wi-Fi"
).grid(row=2, column=0)

tk.Label(
    frame_energie,
    textvariable=wifi
).grid(row=2, column=1)


# =========================
# BOUTON ACTUALISER
# =========================

bouton_actualiser = tk.Button(
    frame_middle,
    text="Actualiser",
    command=maj_donnees
)

bouton_actualiser.grid(
    row=2,
    column=0,
    columnspan=2
)


# =========================
# ÉTAT GLOBAL
# =========================

frame_bottom = tk.Frame(
    root,
    bg="red",
    height=50
)

frame_bottom.pack(
    fill="x"
)

def update_etat_global_color(*args):
    valeur = etat_global.get()
    if valeur == "✅ OK – circulation normale":
        color = "green"
    elif valeur == "⚠️ Alerte":
        color = "orange"
    else:
        color = "red"
    etat_global_label.config(fg=color)

etat_global_label = tk.Label(
    frame_bottom,
    textvariable=etat_global,
    font=("Arial", 16),
    fg="green"
)
etat_global_label.pack()

etat_global.trace_add("write", update_etat_global_color)
update_etat_global_color()


# =========================
# LANCEMENT
# =========================

maj_donnees()

root.mainloop()

````
